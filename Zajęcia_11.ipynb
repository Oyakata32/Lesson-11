```python
#zadanie 15

import pandas as pd

file_path = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
chunksize = 50

total_passengers = 0
sum = 0.0
count = 0
survived_sum = 0

for batch in pd.read_csv(file_path, chunksize=chunksize):

    # Liczba pasażerów
    total_passengers += len(batch)

    # Średni wiek
    valid_age = batch["Age"].dropna()
    sum += valid_age.sum()
    count += valid_age.count()

    # Liczba osób, które przeżyły
    survived_sum += batch["Survived"].sum()

avg_age = sum / count
survival_rate = survived_sum / total_passengers

print("=== Batch processing ===")
print("Total passengers:", total_passengers)
print("Average age:", round(avg_age, 4))
print("Survival rate:", round(survival_rate, 4))

df = pd.read_csv(file_path)
full_total_passengers = len(df)
full_avg_age = df["Age"].mean()
full_survival_rate = df["Survived"].mean()

print("\n=== Full DataFrame ===")
print("Total passengers:", full_total_passengers)
print("Average age:", round(full_avg_age, 4))
print("Survival rate:", round(full_survival_rate, 4))

print("\n=== Comparison ===")
print("Total passengers match:",
      total_passengers == full_total_passengers)

print("Average age match:",
      abs(avg_age - full_avg_age) < 1e-9)

print("Survival rate match:",
      abs(survival_rate - full_survival_rate) < 1e-9)
```
